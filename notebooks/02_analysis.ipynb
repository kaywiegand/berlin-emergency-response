```python
import duckdb
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go

# 1. Verbindung zur DuckDB herstellen (read-only)
con = duckdb.connect('../data/berlin_emergency.duckdb', read_only=True)

# 2. Daten laden
df = con.sql("SELECT * FROM fct_missions_daily ORDER BY mission_date ASC").df()
con.close()

# Erste Zeilen anzeigen
display(df.head())
```

```python
# Interaktiver Plot: Tägliche Einsätze vs. 7-Tage-Durchschnitt
fig = px.line(
    df, 
    x='mission_date', 
    y=['total_missions', 'total_missions_7d_avg'],
    title='Berliner Notfalleinsätze: Tägliche Einsätze & 7-Tage-Trend',
    labels={'value': 'Anzahl Einsätze', 'mission_date': 'Datum', 'variable': 'Metrik'}
)

# Design anpassen
fig.update_layout(
    template='plotly_white',
    hovermode='x unified',
    legend=dict(orientation='h', yanchor='bottom', y=1.02, xanchor='right', x=1)
)

fig.show()
```
